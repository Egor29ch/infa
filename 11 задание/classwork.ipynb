```python
from math import *

A = 910
i = ceil(log2(A))
k = 0

for N in range(1, 1000):
	V1 = ceil((N * i) / 8)
	V1500 = V1 * 1500
	if V1500 <= 780 * 1024:
		k = N
print(k)
```
