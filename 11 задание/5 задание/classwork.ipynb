```python
for N in range(5, 100):
	N_2 = format(N, 'b')
	s = ''
	if N % 2 == 0:
		for i in range(len(N_2) - 1):
			if N_2[i] == '1':
				s += '11'
	elif N % 2 == 2:
		for i in range(len(N_2) - 1):
			if N_2[i] == '1':
				s += '11'
			else:
				s += '0'
	elif N % 2 != 0:
		for i in range(len(N_2) - 1):
			if N_2[i] == '0':
				s += '00'
			else:
				s += '1'
	R = int(s, 2)
	if R > 70:
		print(R)
		break
			
			
		
```

```python
for N in range(5, 200):
	N_2 = format(N, 'b')
	s = ''
	if N % 4 == 0:
		s += N_2 * 2
	else:
		s += N_2
		for i in range(len(N_2) - 1, -1, -1):
			s += N_2[i]
	R = int(s, 2)
	if R >= 544:
		print(N)
		break
```

```python
def to_4(n):
	n_s = ''
	while n > 0:
		n_s = str(n % 4) + n_s
		n //= 4
	return n_s

a = []

for N in range(1, 1000):
	n_s = to_4(N)
	if N % 4 == 0:
		n_s += n_s[-2:]
	else:
		n_s += to_4(N % 4 * 2)
	r = int(n_s, 4)
	if r < 261:
		a.append(N)
print(max(a))
```
