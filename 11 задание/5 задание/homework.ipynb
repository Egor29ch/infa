```python
# 2 вариант
max_R = -1
for N in range(5, 500):
	N_s = format(N, 'b')
	s = ''
	if N % 3 == 0:
		s += N_s + N_s[-3:]
	else:
		s += N_s
		ost_N = format((N % 3 - 1) * 3, 'b')
		s += ost_N
	R = int(s, 2)
	if R < 416 and R > max_R:
		max_R = R
print(max_R)


```

```python
# 4 вариант
min_K = 100000
for K in range(1000, 10000):
	n_1 = K % 10
	n_2 = K % 100 // 10
	n_3 = K % 1000 // 100
	n_4 = K // 1000
	S = n_1 + n_2 + n_3 + n_4
	M = max(n_1, n_2, n_3, n_4)
	N = min(n_1, n_2, n_3, n_4)
	P1 = S - M
	P2 = S - N
	if P1 > P2:
		L = int(str(P1) + str(P2))
		if L == 2013 and K < min_K:
			min_K = K
	else:
		L = int(str(P2) + str(P1))
		if L == 2013 and K < min_K:
			min_K = K
		 
print(min_K)


	
```

```python
# 6 вариант
max_N = -1
for N in range(5, 200):
    N_s = format(N, 'b')
    s = ''
    for i in range(len(N_s)):
        if N % 2 == 0:
            if N_s[i] == '1':
                s += '11'
            else:
                s += '0'
        else:
            if N_s[i] == '0':
                s += '00'
            else:
                s += '1'
    R = int(s, 2)
    if R < 70 and R != N and N > max_N:
        max_N = N
print(max_N)
```

```python
# 10 вариант
a = []
def to_4(n):
	n_4 = ''
	while n > 0:
		n_4 = str(n % 4) + n_4
		n //= 4
	return n_4


for n in range(1, 1000):
	n_4 = to_4(n)
	if n % 4 == 0:
		n_4 += n_4[-2:]
	else:
		n_4 += to_4(n % 4 * 2)
	r = int(n_4, 4)
	if r < 369:
		a.append(n)
print(max(a))
```
