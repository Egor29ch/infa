```python
def F(n):
	if n == 1:
		return 5
	elif n > 1:
		return 2 * n + 1 + F(n - 1)
print(F(2024) - F(2022))
```

```python
def F(n):
	if n == 1:
		return 1
	elif n == 2:
		return 2
	elif n > 2:
		return n * (n - 1) + F(n - 1) - F(n - 2)
	
print(F(2024) + F(2020) - F(2019))
```

```python
def F(n):
	if n <= 1:
		return 0
	elif n > 1 and n % 2 != 0:
		return F(n - 1) + (3 * (n ** 2))
	elif n > 1 and n % 2 == 0:
		return n // 2 + F(n - 1) + 2 
print(F(79))
```
