```python
from itertools import *

k = 0 

for i in product(sorted('СБОРНИК'), repeat = 6):
	w = ''.join(i)
	if w[0] != 'Р':
		if w.count('Б') == 2 and w.count('К') <= 1:
			k += 1
print(k)
```

```python
def F(n):
	if n == 1:
		return 1
	elif n > 1:
		return n - 1 + F(n - 1)

print(F(2024) - F(2022))
```

```python
f = open('17 (2).txt')
a = []
k = 0
max_sum = -1000000
sp_29 = []

for el in f:
	a.append(int(el))

def five_n(n):
	if len(str(abs(n))) == 5:
		return 1
	else:
		return 0

for j in a:
	if abs(j) % 100 == 29:
		sp_29.append(j)

for i in range(len(a) - 2):
	if five_n(a[i]) + five_n(a[i + 1]) + five_n(a[i + 2]) == 2:
		if a[i] + a[i + 1] + a[i + 2] <= max(sp_29):
			k += 1
			max_sum = max(a[i] + a[i + 1] + a[i + 2], max_sum)
print(k, max_sum)


```
