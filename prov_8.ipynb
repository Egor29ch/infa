```python
k = 0

for i in product('01234567', repeat = 5):
	w = ''.join(i)
	if w[0] != '0':
		if w.count('6') == 1:
			if all(d + '6' not in w and '6' + d not in w for d in '1357'):
				k += 1
print(k)
```

```python
k = 0

for i in product(sorted('ВИШНЯ'), repeat = 6):
	w = ''.join(i)
	if w[0] != 'Ш':
		if w[-1] != 'Я'and w[-1] != 'И':
			k += 1
print(k)
```

```python
k = 0
for i in product(sorted('ГЕПАРД'), repeat = 3):
	w = ''.join(i)
	k += 1
	if w[0] == 'Р':
		break
print(k)
```

```python
k = 0
sp_max_k = []
for i in product(sorted('СТРОКА'), repeat = 5):
	w = ''.join(i)
	k += 1
	if k % 2 == 0:
			if (w[0] != 'Р' or w[0] != 'С' or w[0] != 'Т') and w.count('А') == 1:
				sp_max_k.append(k)
print(max(sp_max_k))

```

```python
k = 0
count = 0
for i in product(sorted('АЛГОРИТМ'), repeat = 5):
	w = ''.join(i)
	k += 1
	if k % 2 == 0:
		if w[0] != 'Л' and w.count('И') >= 2:
			count += 1
print(count)

```
