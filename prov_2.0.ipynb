```python
from itertools import *
```

```python
k = 0
for i in product(sorted("ВАЛИК"), repeat= 6):
	w = "".join(i)
	k += 1
	if w.count("А") <= 2 and w.count("В") == 2 and w.count("И") == 0:
		print(k)
		break
```

```python
k = 0
for i in product("0123", repeat = 3):
	w = "".join(i)
	if w[0] != "0":
		if w[0] > w[1] > w[2]:
			k += 1
print(k)
```

```python
k = 0

for i in product("1234", repeat= 3):
	w = "".join(i)
	if w.count("2") == 1:
		for j in "134":
			if w.count(j) == 1 or w.count(j) == 0:
				k += 1
				
print(k)
```

```python
k = 0

for i in product(sorted("ПОЛЬЗА"), repeat= 6):
	w = "".join(i)
	k += 1
	if w.count("Ь") <= 1 and w.count("З") <= 2 and w.count("А") == 1:
		print(k)
		break
```

```python
k = 0 

for i in product("0123", repeat = 5):
	w = "".join(i)
	if w[0] != "0":
		if w.count("3") == 1:
			if "03" not in w and "30" not in w:
				k += 1
print(k)
```
