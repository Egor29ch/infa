```python
print('w x y z')
for w in 0,1:
	for x in 0,1:
		for y in 0,1:
			for z in 0,1:
				if (not(not(x) and y) or (x == z) and w) == 0:
					print(w, x, y, z)
```

```python
print('w x y z')
for w in 0,1:
	for x in 0,1:
		for y in 0,1:
			for z in 0,1:
				if (not(y <= not(z <= w)) and (not(z) <= (not(w) == x))) == 1:
					print(w, x, y, z)
```

```python
for N in range(1, 100):
	n_2 = format(N, 'b')
	if len(n_2) % 2 == 0:
		n_2 = n_2[:len(n_2) // 2] + '1' + n_2[len(n_2) // 2:]
	r = int(n_2, 2)
	if r >= 26:
		print(N)
		break
```

```python
from itertools import *

k = 0

for i in product(sorted('СОРНЯК'), repeat= 6):
	w = ''.join(i)
	k += 1
	if w.count('К') <= 3 and w.count('Я') == 2:
		print(k)
		break
```
