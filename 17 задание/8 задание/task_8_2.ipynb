```python
from itertools import *
count = 0
for x in permutations('0123456789', 5):
	w = ''.join(x)
	if w[0] != '0':
		if all(int(w[i]) % 2 == 0 and int(w[i + 1]) % 2 != 0 or int(w[i]) % 2 == 0 and int(w[i + 1]) % 2 != 0 for i range(len(w) - 1)):
			count += 1
print(count)
```
