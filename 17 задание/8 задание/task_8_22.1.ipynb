```python
from itertools import *

count = 0

for i in product('0123456', repeat = 6):
    ch = ''.join(i)
    if ch[0] != '0':
        if ch.count('0') == 1:
            if all(d + '0' not in ch and '0' + d not in ch for d in '246'):
                count += 1

print(count)
```
